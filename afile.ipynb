```python
# load library
import numpy as np
from scipy import sparse

"""
Problem:

Given data with very few nonzero values, you want to efficiently represent it.

"""

# Matrix
matrix = np.array([[0, 0],
                  [0,1],
                  [3,0]])

# Compressed sparse row (CSR) matrix
matrix_sparse = sparse.csr_matrix(matrix)

# View sparse matrix
print(matrix_sparse)
```
